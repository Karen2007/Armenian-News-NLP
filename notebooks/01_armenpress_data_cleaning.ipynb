```python
# Remove prefixes like 'ԵՐԵՎԱՆ, 9 ՀՈՒԼԻՍԻ, ԱՐՄԵՆՊՐԵՍ:'
def remove_armenpress_prefix(text):
    text = re.sub(r'^.+?ԱՐՄԵՆՊՐԵՍ[։:]', '', text).strip()
    return text

df['text'] = df['text'].apply(remove_armenpress_prefix)
```

```python
# Fix issue where the armenian letter 'ֆ' is instead 'ü' or 'ý'
df['text'] = df['text'].str.replace('ý', 'ֆ')
df['text'] = df['text'].str.replace('ü', 'ֆ')

df['title'] = df['title'].str.replace('ý', 'ֆ')
df['title'] = df['title'].str.replace('ü', 'ֆ')
```

```python
# Remove whitespace encodings from articles and titles
df['text'] = df['text'].str.replace('\xa0', ' ')
df['text'] = df['text'].str.replace('\x7f', '')
df['text'] = df['text'].str.replace('\n', ' ')
df['text'] = df['text'].str.replace('\t', ' ')

df['title'] = df['title'].str.replace('\xa0', ' ')
df['title'] = df['title'].str.replace('\x7f', '')
df['title'] = df['title'].str.replace('\n', ' ')
df['title'] = df['title'].str.replace('\t', ' ')
```

```python
# Check if removing was successful
print(df['text'].str.contains('ý').sum())
print(df['text'].str.contains('ü').sum())
```

```python
# Check if whitepsaces got removed
print(df['text'].str.contains('\n').sum())
print(df['text'].str.contains('\t').sum())
print(df['text'].str.contains('\xa0').sum())
print(df['text'].str.contains('\x7f').sum())
```

```python
# Replace multiple spaces with just one
df['text'] = df['text'].str.replace(r'\s{2,}', ' ', regex=True)
df['title'] = df['title'].str.replace(r'\s{2,}', ' ', regex=True)
```

```python
# Merge categories to general names
category_mapping = {
    'Սոցիալական հարցեր': 'Society',
    'Հասարակական': 'Society',

    'Տնտեսություն': 'Economy',
    'Ֆինանսներ': 'Economy',

    'Միջազգային': 'International',

    'Քաղաքականություն': 'Politics',
    'Պաշտոնական': 'Politics',
    'Ռազմական': 'Politics',
    'Իրավունք': 'Politics',

    'Սպորտ': 'Sport',
    'Ֆուտբոլ': 'Sport',

    'Մշակույթ': 'Culture',
    'Շոու բիզնես': 'Culture',
    'Life': 'Culture',
    'Կրթություն': 'Culture',

    'Բարձր տեխնոլոգիաներ' : 'Tech',
    'Ինովացիա' : 'Tech'
}

df['category'] = df['category'].map(category_mapping)
df = df.dropna(subset=['category'])  # Drop anything not in the mapping
print(df['category'].value_counts())
```
